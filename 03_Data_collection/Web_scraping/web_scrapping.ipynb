# Working with Requests

```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"

res = requests.get(URL)
```

```python
print(res.status_code) # to get the status code

print(res.text)  # it gives the html content of the page as string

print(res.content)   # it gives the raw content of the response

print(res.headers)  # gives the metadata like content type, server info
```

we need to store the responses in files as there will be many responses and they will not be easy to handle in variables.

here is how we will write in file :

```python
with open("Scrapped_files/data1.html","w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup
```

```python
with open("Scrapped_files/data1.html","r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")
```

#### Methods of BeautifulSoup

```python
# find()    ->> gives the first occuring of matching element

soup.find("h3")
```

```python
# find_all()   -->> gives all occurings of matching element 

all_h3 = soup.find_all("h3")
type(all_h3)     # its the beautifulSoup type just like the lists

for h3 in all_h3:
    print(h3)

# to get the text which is inside the tags we use get_text()
for h3 in all_h3:
    print(h3.get_text(strip = True))
```

```python
# find_parent()  -->> it gives the parent of selected element

for h3 in all_h3 :
    print(h3.find_parent("div")["class"])


# find_next()  -->> it gives the next element of the selected element

# find_previous()  -->> it gives the previous element of the selected element

## above two are used to move forward or backward while working on these elements
```

### css selectors and attribute

```python
# select()    -->> 

for h3 in all_h3 :
    print(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))
    

# select_one()    -->>

for h3 in all_h3 :
    print(h3.find_next("div").select_one("span.country-population").get_text(strip=True))
```
