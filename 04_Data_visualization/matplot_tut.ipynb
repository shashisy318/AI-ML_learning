```python
# working on real data

# oscar winning movies analysis
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232] # in $M

#plot 
plt.plot(years,oscar_revenue)

# title
plt.title("Oscar Movies Revenue in each Year")

# xlabel  ,  ylabel
plt.xlabel("Years") 
plt.ylabel("Revenue  (in $M)")
```

## Different Plots

    -> plt.plot(x,y) - line plot
    
    -> plt.bar(x,height) - vertical bar chart
    
    -> plt.barh(x,width) - horizontal bar chart
    
    -> plt.scatter(x,y) - scatter plot
    
    -> plt.pie(data) - pie chart
    
    -> plt.hist(data,bins) - Histogram
    
    -> plt.boxplot(data) - Box plot


# Bar Charts

```python
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008,2009,2010,2011,2012]
revenue = [1005,170,427,133,232] # in $M

plt.bar(years,revenue , color="Orange")
plt.title("Oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue  (in $M)")

## plt.text(x-axis, y-axis, text, alignment)  # used to add text on the plot
for i in range(len(years)):
    plt.text(years[i], revenue[i]+10, str(revenue[i]), ha='center')

## plt.xlim(min,max)   plt.ylim(min,max)   ## used to increase the spacing
plt.ylim(0, max(revenue)+100)
```

* Multiple datasets on Bar Chart

```python
import numpy as np
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005,170,427,133,232] # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378,2788,829,185,275]   # in $M

years = [2008,2009,2010,2011,2012]

x = np.arange(len(years))   # this generates an array of length of years and used to send the position on x-axis 
width = 0.4    # this sets the width of the bar to 0.4

plt.bar(x-width/2, oscar_revenue, width, label = "Oscar Movies")  # x-width/2 is the starting point for the bar
plt.bar(x+width/2, non_oscar_revenue, width, label = "Non-Oscar Movies")
plt.title("Oscar Movies vs Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.xticks(x,years)  # this fn replaces the x-axis labels
```

# Scatter Plots
* it shows the relationship between two variable by plotting points 

-> for correlated data

-> Detect outliers

-> Forms cluster

```python
# scatter plot

people =["Person A","Person B","Person C","Person D","Person E",
         "Person F","Person G","Person H","Person I","Person J"]
age = [22,25,30,35,40,45,50,55,60,65]
bp = [110,115,120,120,125,130,135,123,145,150]

colors = ["green" if x<135 else "red" for x in bp]

plt.scatter(age,bp,color = colors, s = bp, alpha = 0.5, marker ='o')  # can also use cmap =orRd and c = bp for better color representation
plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar(label = "bp")

for i in range(len(people)):
    plt.annotate(people[i],xy= (age[i],bp[i]), xytext = (age[i],bp[i]-2))

plt.xlim(min(age)-2,max(age)+10)
plt.ylim(min(bp)-5,max(bp)+5)
```

* Multiple Dataset on scatter plot

```python
#summer
winter_temp  = [5,2,10,0,7]
winter_humidity = [80,75,65,85,70]
#winter
summer_temp = [25,30,28,35,27]
summer_humidity = [60,50,55,45,65]

plt.scatter(winter_temp,winter_humidity, label = "Winter")
plt.scatter(summer_temp,summer_humidity , label = "Summer")

plt.title("Summer vs Winter Data")
plt.xlabel("Temperature (*C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

# Pie Charts

* to show relative proportion of a whole.
* prefer maximum of 6 slices.

```python
# Pie chart
expenses = ["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts = [500,150,120,100,50]

explodes = [0,0,0,0.1,0]

plt.pie(amounts, labels = expenses , 
        autopct="%1.1f%%",   # this shows the exact percentage of slice
        wedgeprops = {   ## this actully is for style of boundaries
            "edgecolor":"black",
            "linewidth":1,
            "linestyle":"--",
            # "fill": False
        },
        explode = explodes,   # this explodes out the slice from the pie as much value is passed
        startangle = 360,  # this rotates the pie
        shadow = True
       )
```
